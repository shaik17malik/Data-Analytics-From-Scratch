```python
# Cheatsheet code

# ============ DAY 2 CHEATSHEET ============

# --- SETUP ---
import pandas as pd
import numpy as np

df = pd.read_csv('your_file.csv')
df_clean = df.copy()          # ALWAYS copy first!

# --- FIND PROBLEMS ---
df.isnull().sum()             # Count missing values per column
df.duplicated().sum()         # Count duplicate rows
df.describe()                 # Look at min/max to spot invalid values

# --- FIX MISSING VALUES ---
# Option A: Fill
df['col'] = df['col'].fillna(value)     # Fill with a specific value
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())  # Fill with average
df['Name'] = df['Name'].fillna('Unknown')             # Fill with text

# Option B: Drop
df = df.dropna()                        # Drop rows with ANY missing value

# --- REMOVE DUPLICATES ---
df = df.drop_duplicates()               # Remove duplicate rows

# --- FIX INVALID VALUES ---
# Pattern: df.loc[condition, 'col'] = new_value

df.loc[df['Age'] == 150, 'Age'] = 19              # Replace specific value
df.loc[df['Marks'] < 0, 'Marks'] = np.nan         # Turn invalid into missing
df.loc[df['Att'] > 100, 'Att'] = 100              # Cap at a max

# --- VERIFY & SAVE ---
df.isnull().sum()                # Should be all zeros
df.duplicated().sum()            # Should be 0
df.to_csv('clean_data.csv', index=False)

# --- REMEMBER ---
# 1. Always copy first: df_clean = df.copy()
# 2. Few missing → drop. Many → fill.
# 3. You decide what values are "invalid."
```

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("/content/student_performance_dataset (2).csv")

# Calculate average marks
average_marks = df["Marks"].mean()

print(f"Average Marks of All Students: {average_marks:.2f}")
```

```python
# Filter students with marks above 75
high_performers = df[df["Marks"] > 75]

# Count high-performing students
high_performer_count = len(high_performers)

print(f"Number of students scoring above 75: {high_performer_count}")
high_performers
```

```python
import matplotlib.pyplot as plt
plt.scatter(df["Attendance_Percentage"], df["Marks"])
plt.xlabel("Attendance_Percentage")
plt.ylabel("Marks")
plt.title("Marks vs Attendance_Percentage")
plt.grid(True)
```

```python
# Calculate Pearson correlation between attendance and marks
correlation = df["Attendance_Percentage"].corr(df["Marks"])

print(f"Correlation between Attendance and Marks: {correlation:.2f}")
```

```python
# True for marks of 50 or higher; otherwise False
df["Pass"] = df["Marks"] >= 50
```

```python
df["Pass"] = df["Marks"].apply(lambda x: True if x >= 50 else False)
```

```python
# Summary counts
passes = df["Pass"].sum()
fails = (~df["Pass"]).sum()

print(f"Passes: {passes}")
print(f"Fails: {fails}")

# Histogram of marks
plt.figure(figsize=(8, 5))
plt.hist(df["Marks"], bins=6, edgecolor="black", color="skyblue")

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.grid(True, axis="y", linestyle="--", alpha=0.7)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv("/content/student_performance_dataset (2).csv")

# Create Pass status: marks >= 50 are passing
df["Pass"] = df["Marks"] >= 50

# Summary counts
passes = df["Pass"].sum()
fails = (~df["Pass"]).sum()

print(f"Passes: {passes}")
print(f"Fails: {fails}")

# Histogram of marks
plt.figure(figsize=(8, 5))
plt.hist(df["Marks"], bins=6, edgecolor="black", color="skyblue")

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.grid(True, axis="y", linestyle="--", alpha=0.7)

plt.show()
```
